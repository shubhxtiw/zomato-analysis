```python
country_counts = df['country'].value_counts()
country_counts.head(10)
```

```python
plt.figure(figsize=(10,6))

sns.barplot(
    x=country_counts.head(10).values,
    y=country_counts.head(10).index
)

plt.title("Top Countries by Restaurant Count")
plt.xlabel("Number of Restaurants")
plt.ylabel("Country")

plt.show()
```

```python
city_counts = df['city'].value_counts().head(10)
city_counts
```

```python
plt.figure(figsize=(10,6))
sns.barplot(
    x=city_counts.values,
    y=city_counts.index
)

plt.title("Top 10 Cities with Most Restaurants")

plt.show()
```

```python
plt.figure(figsize=(8,5))

sns.histplot(
    df['aggregate_rating'],
    bins=20,
    kde=True
)

plt.title("Restaurant Rating Distribution")

plt.show()
```

```python
df['aggregate_rating'].describe()
```

```python
df['Price range'].value_counts()
```

```python
plt.figure(figsize=(7,5))

sns.countplot(
    x='Price range',
    data=df
)

plt.title("Price Range Distribution")

plt.show()
```

```python
plt.figure(figsize=(10,5))

sns.boxplot(
    x=df['avg_cost_two']
)

plt.title("Average Cost for Two")

plt.show()
```

```python
df['primary_cuisine'].value_counts().head(10)
```

```python
top_cuisines = df['primary_cuisine'].value_counts().head(10)

plt.figure(figsize=(10,6))

sns.barplot(
    x=top_cuisines.values,
    y=top_cuisines.index
)

plt.title("Top 10 Cuisines")

plt.show()
```

```python
df['has_online_delivery'].value_counts()
```

```python
sns.countplot(
    x='has_online_delivery',
    data=df
)

plt.title("Online Delivery Availability")

plt.show()
```

```python
df['Has Table booking'].value_counts()
```

```python
sns.countplot(
    x='Has Table booking',
    data=df
)

plt.title("Table Booking Availability")

plt.show()
```

```python
plt.figure(figsize=(10,6))

sns.scatterplot(
    x='avg_cost_two',
    y='aggregate_rating',
    data=df
)

plt.title("Rating vs Average Cost")

plt.show()
```
